```python
import json

with open("COMBINED-FINAL-DEDUPED-CLEAN2.json", "r", encoding="utf-8") as origin:
    origin_jsons = json.load(origin)

with open("COMBINED-FINAL-DEDUPED-CLEAN3.json", "w") as dest:

    processed_jsons = []
    
    for origin_json in origin_jsons:

        # SPECIAL CASE FOR DRESSES
        if "dress" in origin_json['name'].lower() or "dress" in origin_json['description'].lower():

            # SWATH 1: 189 to 332 for men
            # SWATH 2: 1048 to 1125 for men
            if ((origin_json['ID'] >= 189 and origin_json['ID'] <= 332)
                or (origin_json['ID'] >= 1048 and origin_json['ID'] <= 1125)):
                origin_json['gender'] = 'Men'
            else:
                origin_json['gender'] = "Women"
            print(origin_json)
        
        # GENERAL CASE
        if origin_json['gender'] == 'm':
            origin_json['gender'] = "Men"
        elif origin_json['gender'] == 'w':
            origin_json['gender'] = "Women"

        processed_jsons.append(origin_json)
    
    dest.write(json.dumps(processed_jsons, indent=4))
```
